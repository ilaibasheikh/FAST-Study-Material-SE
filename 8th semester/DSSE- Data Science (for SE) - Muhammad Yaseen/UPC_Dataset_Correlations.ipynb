# Chi-Square ($\chi^2$) Measure

```python
ct = pd.crosstab(df['Sector'], df['ApplicationType'])
print(ct)
```

```python
chi2, p, dof, expected = chi2_contingency(ct)

print("Chi-square:", chi2)
print("p-value:", p)
print("DOF:", dof)
print("Expected:\n", expected)
```

Cramer's $V$

```python
n = ct.sum().sum()
r, c = ct.shape

cramers_v = np.sqrt(chi2 / (n * (min(r-1, c-1))))

print("Cramer's V:", cramers_v)
```

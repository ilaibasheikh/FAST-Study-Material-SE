```python
import socket

s = socket.socket()
s.connect(('localhost', 8899))

server_message = s.recv(1024).decode("utf-8")
print("Server says:", server_message)

while True:
    message_to_server = input("Enter message to send to server (type 'quit' to exit): ")
    s.send(bytes(message_to_server, 'utf-8'))
    if message_to_server.lower() == "quit":
        server_response = s.recv(1024).decode('utf-8')
        print("Server responded:", server_response)
        break
    server_response = s.recv(1024).decode('utf-8')
    print("Server responded:", server_response)
s.close()
```

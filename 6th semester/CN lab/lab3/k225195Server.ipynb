```python
import socket
s = socket.socket()
print("Socket created")
s.bind(('localhost', 9999))
s.listen(5)
print("Waiting for connection")
while True:
    c, addr = s.accept()
    print("Got connection from", addr)
    c.send(bytes("Thankyou for connecting", 'utf-8'))  # Encoding done correctly here
    c.close()
```

```python
from socket import *

import time

startTime=time.time()

if __name__=='__main__':
    target=input("Enter hostname to be scanned: ")
    target_ip=gethostbyname(target)


    for i in range(50,500):
        s=socket(AF_INET,SOCK_STREAM)

        conn=s.connect_ex((target_ip,i))

        if(conn==0):
            print("Port %d is OPEN" % (i,))
        s.close()

print("Time Taken in Seconds: ",time.time()-startTime)
print("Time Taken in Minutes: ",(time.time()-startTime)/60)



```

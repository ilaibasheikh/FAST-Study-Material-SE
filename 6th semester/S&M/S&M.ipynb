```python
def prob_calc(n,event):
    counter=0
    for i in range(n):
        die_throw=randint(1,6) 
        if die_throw ==event:
            counter+1
    return counter/n
```

```python
n=10
counter=0
event=2
observed = []
for i in range(n):
    die_throw= randint(1,6)
    if die_throw== event:
        counter+=1 
        observed.append(die_throw)

print("Probability of the event is",counter/n)
```

```python
#computing moving average
```

```python
cum_observed = np.cumsum(observed)
moving_avg = []
for i in range(len(cum_observed)):
    moving_avg.append(cum_observed[i]/(i+1))### Part 2: Computing the moving average.
```

```python
#Making plot
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x= np.arange(0, len(moving_avg), 1) # x-axis 
p =[3.5 for i in range(len(moving_avg))] #Line
plt.xlabel('Iterations', size= 20) 
plt.ylabel('Probability', size=20)
plt.plot(x, moving_avg)
plt.plot(x, p, linewidth= 2, color='black')
plt.show()
```

```python
from random import choice 
#from statistics import mean
n=10
observed =[]
for i in range(n):
    outcome =choice(['Head', 'Tail'])
    if outcome =='Head':
        observed.append(1)
    else:
        observed.append(0)
print('Prob', round(sum(observed)/n, 2))
```

```python
#Computing moving average
```

```python
cum_observed= np.cumsum(observed)
moving_avg = [] 
for i in range(len(cum_observed)):
    moving_avg.append(cum_observed[i]/(i+1))
```

```python
#Making plot
```

```python
x= np.arange(0, len(moving_avg), 1) # x-axis 
p= [0.5 for i in range(len(moving_avg))] # Line
plt.xlabel('Iterations', size=20) 
plt.ylabel('Probability', size=20) 
plt.plot(x, moving_avg) 
plt.plot(x, p, linewidth=2, color='black')
plt.show()
```

```python
def rollDie(): 
    #returns a random int between 1 and 6 
    return random.choice ([1,2,3,4,5,6])
    
def testRoll(n = 10):
    result=''
    for i in range(n):
        result =result+ str(rollDie())
    print(result)
    
def runsin(goal, numTrials): 
    total =0
    for i in range(numTrials):
        result =''
    for j in range(len(goal)):
        result +=str(rollDie())
    if result == goal: 
        total+= 1
    print('Actual probability =', round(1/(6**len(goal)), 8)) 
    estProbability =round(total/numTrials, 8)
    print('Estimated Probability=', round(estProbability, 8))
```
